#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)

- Comment your code throughout

```r
basketball_tidy <- basketball_data |> # The dataset we are tidying.
    # Add a new column labelling teams who have won one or more games as "Yes" and teams that won zero games as "No" and converts it into a factor.
    mutate(won_tourney_game = as_factor(if_else(tourney_wins > 0, "Yes", "No"))) |>
    # Select only the relevant columns (whether they won a game, regional win percentage, and seed) for our analysis.
    # Tournament columns are left out because they're only known after the tournament and would thus make our model seem more accurate than it is. 
    select(won_tourney_game, reg_wins_pct, seed) |>
    # Filter out the teams with a seed of 0 since the analysis would treat them as being better than a first seed team, which could mess up the classification model.
    filter(seed > 0)

# Display the first 6 rows of the tidied data
head(basketball_tidy)
```

#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
basketball_summary <- basketball_tidy |>
    # Select the numeric variables to be summarized
    select(reg_wins_pct, seed) |>
    # Convert the selected columns into a longer format, with variable names
    # stored in "variable" and their corresponding values stored in "value"
    pivot_longer(reg_wins_pct:seed, names_to = "variable", values_to = "value") |>
    # Group the data by variable so that statistics are calculated separately
    group_by(variable) |>
    # Calculate the mean, minimum, and maximum for each variable,
    # excluding missing values from these calculations
    summarize(
        mean_value = mean(value, na.rm = TRUE),
        min_value = min(value, na.rm = TRUE),
        max_value = max(value, na.rm = TRUE)
    )

# Display the summary table
basketball_summary
```

<div style="background-color: #f0f0f0; padding: 10px;">
    
#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).
- Show the outputted table. 
</div>

```r
basketball_na <- basketball_tidy |>
    # Replace each value with TRUE if it is missing (NA), and FALSE otherwise
    mutate(across(won_tourney_game:seed, is.na)) |>
    # Sum the TRUE values in each column to get the total number of missing values
    summarize(across(won_tourney_game:seed, sum))

# Display the number of missing values in each column
basketball_na
```

Visualization 1: This visualization compares teams' regional win percentages and tournament seeds, with points coloured by whether they won at least one tournament game. Teams with lower seed numbers are more likely to win at least one game, while teams with higher seed numbers are more likely to lose their first game. However, the two outcome groups overlap considerably among middle seeds. Within individual seeds, regional win percentage appears to have a weaker relationship with tournament success. This suggests that seed may be a stronger predictor than regional win percentage, although both variables may contribute to the model.

```r
# Set the height and width of the plot
options(repr.plot.height = 9, repr.plot.width = 12)

predictors_scatterplot <- ggplot(basketball_tidy, aes(x = as.factor(seed), y = reg_wins_pct, color = won_tourney_game)) +
    # Plot each team as a point, using transparency to reduce overplotting
    geom_point(alpha = 0.7) +
    # Add a title and labels for the axes and outcome categories
    labs(
        x = "Tournament Seed",
        y = "Regional Win Percentage (%)",
        colour = "Won at Least One Tournament Game",
        title = "Regional Win Percentage and Seed by Tournament Outcome"
    ) +
    # Increase the text size for readability
    theme(text = element_text(size = 16)) +
    # Apply the Set2 colour palette to distinguish the outcome categories and make it colorblind friendly
    scale_colour_brewer(palette = "Set2")

# Display the scatter plot
predictors_scatterplot
```

Visualization 2: This visualization shows the proportion of teams that won at least one tournament game for each seed. Displaying proportions rather than raw counts makes comparisons between seeds with different numbers of teams easier. This pattern further reinforces that seed may be a useful predictor of tournament success.

```r
# Set the height and width of the plot
options(repr.plot.height = 9, repr.plot.width = 14)

seed_win_prop <- ggplot(basketball_tidy, aes(x = as.factor(seed), fill = won_tourney_game)) +
    # Display the proportion of teams in each tournament outcome category for each seed
    geom_bar(stat = "count", position = "fill") +
    # Add a title and labels for the axes and fill categories
    labs(
        x = "Tournament Seed",
        y = "Proportion of Teams",
        fill = "Won at Least One Tournament Game",
        title = "Proportion of Teams Winning at Least One Game by Seed"
    ) +
    # Increase the text size for readability
    theme(text = element_text(size = 20)) +
    # Apply the Set2 colour palette to distinguish the outcome categories and make it colorblind friendly
    scale_fill_brewer(palette = "Set2")
                
# Display the proportional bar chart
seed_win_prop
```
